# VIX VECM v1 — le modèle de correction d'erreur bat-il la référence ML en walk-forward ?

**Rôle.** Suite directe de `VIX_VAR_MACRO` : si le test de Johansen y détecte une
cointégration (rang r>0) entre le VIX et les variables macro, ce notebook exploite
explicitement cette relation de long terme via un VECM (Vector Error Correction Model)
pour **prévoir le VIX** et compare sa performance directionnelle, en walk-forward strict,
à la référence établie du projet (GLOBAL RandomForest : F1_dir≈0.610±0.025).

**Principe.** Pour chaque fold walk-forward (mêmes coupures que le reste du projet) :
1. Test de Johansen sur le train uniquement (causal) → rang de cointégration r.
2. Si r=0 : le fold est ignoré pour ce notebook (rien à corriger, cf. `VIX_VAR_MACRO`).
3. Si r>0 : VECM(k_ar_diff, coint_rank=r) estimé sur le train ; sa représentation VAR en
   niveaux équivalente (`var_rep`, cf. Lütkepohl 2005) sert ensuite à produire, pour
   **chaque jour du test**, une prévision à horizon h en repartant des valeurs réelles
   passées jusqu'à ce jour (paramètres figés à l'entraînement, aucune fuite du futur).
4. La prévision de rendement du VIX est convertie en classe (DOWN_FORT/FAIBLE,
   UP_FAIBLE/FORT) avec les **mêmes seuils causaux** que le pipeline ML (`build_target`),
   pour une comparaison rigoureusement à isométrie de méthode.

**Ce que ce notebook ne fait PAS** : pas de sélection de features, pas d'apprentissage —
c'est un modèle économétrique paramétrique pur, avec deux inconnues seulement (l'ordre
de retard p et le rang de cointégration r), à l'opposé du pipeline ML (SHAP + ensembles).
L'objectif est de savoir si la structure de long terme identifiée par Johansen contient,
à elle seule, un signal directionnel comparable à celui trouvé par le ML.

```python
import os, time, json, warnings, random, subprocess
warnings.filterwarnings('ignore')
import numpy as np
import pandas as pd
import pandas_datareader.data as web
from sklearn.metrics import f1_score, accuracy_score
from statsmodels.tsa.vector_ar.var_model import VAR
from statsmodels.tsa.vector_ar.vecm import VECM, coint_johansen

SEED = 42; random.seed(SEED); np.random.seed(SEED)

NOTEBOOK_NAME = 'VIX_VECM'
NOTEBOOK_VERSION = 'v1'

CONFIG = {
    'macro_series': {'T10Y2Y': 'T10Y2Y', 'T10YIE': 'T10YIE', 'NFCI': 'NFCI', 'EFFR': 'EFFR'},
    'start_date': '2003-01-01',
    'flat_thr': 0.003,
    'horizons': [1, 2, 3, 5, 7, 10],
    'n_wf_folds': 5,
    'min_train_frac': 0.40,   # doit matcher VIX_FINAL_FEATURES pour les mêmes folds
    'maxlags': 10,
    'johansen_det_order': 0,  # constante, pas de tendance (cohérent avec VIX_VAR_MACRO)
    'min_train_rows': 250, 'min_test_rows': 20,
}
TARGET_COL = 'VIX_Amplitude_Class'
RESULTS_CSV = 'vix_vecm_results.csv'
GITHUB_REPO = 'LP-D/claude'
FEATURES_BRANCH = 'results/vix-final-features'
RESULTS_BRANCH = 'results/vix-vecm'

# Référence établie (README) pour comparaison directe — GLOBAL RandomForest h=5j N=8 SHAP SMOTE
BASELINE_F1_DIR = 0.610
BASELINE_F1_UP_FORT = 0.359
BASELINE_F1_DOWN_FORT = 0.627

print(f"{NOTEBOOK_NAME} {NOTEBOOK_VERSION} | macro: {list(CONFIG['macro_series'].keys())} | "
      f"horizons={CONFIG['horizons']} | référence à battre: F1_dir={BASELINE_F1_DIR}")
```

```python
# ============================================================
# CHARGEMENT DU DATASET PARTAGÉ (produit par VIX_FINAL_FEATURES) — pour retrouver
# exactement le même VIX_COL, les mêmes dates et les mêmes coupures de fold que la
# référence ML établie du projet (comparaison à isométrie de méthode).
# ============================================================
try:
    from google.colab import userdata
    GITHUB_TOKEN = userdata.get('GITHUB_TOKEN')
except Exception:
    GITHUB_TOKEN = os.environ.get('GITHUB_TOKEN')

if not os.path.exists('vix_final_features.parquet'):
    auth = f"{GITHUB_TOKEN}@" if GITHUB_TOKEN else ""
    url = f"https://{auth}github.com/{GITHUB_REPO}.git"
    workdir = "/content/_vix_features_pull"
    subprocess.run(["rm", "-rf", workdir], check=False)
    clone = subprocess.run(["git", "clone", "--depth", "1", "--branch", FEATURES_BRANCH, url, workdir],
                           capture_output=True, text=True)
    if clone.returncode != 0:
        raise RuntimeError(
            "Impossible de récupérer le dataset partagé depuis "
            f"'{FEATURES_BRANCH}'. As-tu bien exécuté VIX_FINAL_FEATURES.ipynb en premier "
            f"(et poussé son résultat) ? Détail: {clone.stderr[-500:]}")
    subprocess.run(["cp", f"{workdir}/vix_final_features.parquet", "."], check=True)
    subprocess.run(["cp", f"{workdir}/vix_final_features_meta.json", "."], check=True)
    print(f"[PULL OK] Dataset récupéré depuis '{FEATURES_BRANCH}'")
else:
    print("[SKIP] vix_final_features.parquet déjà présent localement")

df_features = pd.read_parquet('vix_final_features.parquet')
with open('vix_final_features_meta.json') as f:
    meta = json.load(f)
VIX_COL = meta['vix_col']
print(f"Dataset: {df_features.shape} | VIX={VIX_COL} | source: {meta['date_min']} → {meta['date_max']}")

all_dates = df_features.dropna(how='all').index.sort_values()
n_obs = len(all_dates)
first_cut = int(n_obs * CONFIG['min_train_frac'])
test_span = (n_obs - first_cut) // CONFIG['n_wf_folds']
FOLD_CUTS = [first_cut + k * test_span for k in range(CONFIG['n_wf_folds'] + 1)]
FOLD_CUTS[-1] = n_obs
for k in range(CONFIG['n_wf_folds']):
    print(f"  Fold {k+1}: train → {all_dates[FOLD_CUTS[k]-1].date()} | "
          f"test {all_dates[FOLD_CUTS[k]].date()} → {all_dates[FOLD_CUTS[k+1]-1].date()}")
```

```python
# ============================================================
# SYSTÈME VAR/VECM : VIX (issu du dataset partagé, pour cohérence avec la
# référence ML) + variables macro FRED, alignées sur le même index.
# ============================================================
system_df = pd.DataFrame(index=df_features.index)
system_df['VIX'] = df_features[VIX_COL]

for name, sid in CONFIG['macro_series'].items():
    try:
        s = web.DataReader(sid, 'fred', CONFIG['start_date']).squeeze()
        system_df[name] = s.reindex(system_df.index, method='ffill')
    except Exception as e:
        print(f"  [WARN] FRED {sid}: {str(e)[:100]}")

VAR_COLS = list(system_df.columns)
VIX_IDX = VAR_COLS.index('VIX')
system_df = system_df.ffill().dropna()
print(f"Système: {system_df.shape} | colonnes: {VAR_COLS} | {system_df.index.min().date()} → "
      f"{system_df.index.max().date()}")
```

```python
def build_target(vix_series, horizon, split_idx):
    vix = vix_series.ffill().bfill(); vix_tr = vix.iloc[:split_idx]
    calm_thr = vix_tr.quantile(0.33); stress_thr = vix_tr.quantile(0.67)
    regime = pd.Series('NORMAL', index=vix.index)
    regime[vix < calm_thr] = 'CALM'; regime[vix >= stress_thr] = 'STRESS'
    ret = (vix.shift(-horizon) / vix) - 1
    flat = ret.abs() < CONFIG['flat_thr']
    ret = ret.loc[~flat].dropna(); reg_r = regime.reindex(ret.index)
    cut_date = vix.index[min(split_idx, len(vix) - 1)]
    ret_tr = ret.loc[ret.index < cut_date]; reg_tr = reg_r.loc[ret_tr.index]
    thr = {}
    for reg in ['CALM', 'NORMAL', 'STRESS']:
        sub = ret_tr[reg_tr == reg]
        thr[reg] = (sub.quantile(0.25) if len(sub) >= 20 else ret_tr.quantile(0.25),
                    sub.quantile(0.75) if len(sub) >= 20 else ret_tr.quantile(0.75))
    thr['GLOBAL'] = (ret_tr.quantile(0.25), ret_tr.quantile(0.75))
    def classify(r, reg):
        q25, q75 = thr.get(reg, (0, 0))
        if r < q25: return 0
        if r < 0:   return 1
        if r < q75: return 2
        return 3
    target = pd.Series([classify(r, reg_r[i]) for i, r in ret.items()], index=ret.index, name=TARGET_COL)
    return target, reg_r, thr

def metrics(y_true, y_pred):
    dm = {0: 'DOWN', 1: 'DOWN', 2: 'UP', 3: 'UP'}
    yd_t = [dm[y] for y in y_true]; yd_p = [dm[y] for y in y_pred]
    m = {'F1_4cls': round(f1_score(y_true, y_pred, average='macro', zero_division=0), 4),
         'Acc_dir': round(accuracy_score(yd_t, yd_p), 4),
         'F1_dir': round(f1_score(yd_t, yd_p, average='macro', zero_division=0), 4)}
    ui = [i for i, y in enumerate(y_true) if dm[y] == 'UP']
    di = [i for i, y in enumerate(y_true) if dm[y] == 'DOWN']
    if len(ui) >= 10:
        yt = ['FORT' if y_true[i] == 3 else 'FAIBLE' for i in ui]
        yp = ['FORT' if y_pred[i] == 3 else 'FAIBLE' for i in ui]
        m['F1_UP_FORT'] = round(f1_score(yt, yp, pos_label='FORT', average='binary', zero_division=0), 4)
    else: m['F1_UP_FORT'] = np.nan
    if len(di) >= 10:
        yt = ['FORT' if y_true[i] == 0 else 'FAIBLE' for i in di]
        yp = ['FORT' if y_pred[i] == 0 else 'FAIBLE' for i in di]
        m['F1_DOWN_FORT'] = round(f1_score(yt, yp, pos_label='FORT', average='binary', zero_division=0), 4)
    else: m['F1_DOWN_FORT'] = np.nan
    return m

def classify_return(r, reg, thr):
    q25, q75 = thr.get(reg, (0, 0))
    if r < q25: return 0
    if r < 0:   return 1
    if r < q75: return 2
    return 3

def var_level_forecast(coefs, const, y_hist_window, h):
    # coefs: (p, K, K) ; const: (K,) ; y_hist_window: (p, K) réel, du plus ancien au plus
    # récent. Récursion identique à la représentation VAR en niveaux (validée contre
    # VECMResults.predict() — cf. Lütkepohl 2005, pp. 269-304).
    p = coefs.shape[0]
    hist = [row.copy() for row in y_hist_window]
    for _ in range(h):
        yhat = const.copy()
        for i in range(p):
            yhat = yhat + coefs[i] @ hist[-(i + 1)]
        hist.append(yhat)
    return hist[-1]

print("Helpers OK (build_target, metrics, classify_return, var_level_forecast)")
```

### VECM et représentation VAR en niveaux équivalente

Sous forme VECM : `Δyt = Γ1·Δyt-1 + ... + Γp-1·Δyt-p+1 + α·β'·yt-1 + ut`, avec
`α·β' = Π = -(I_K - A1 - ... - Ap)` de rang r (le rang de cointégration trouvé par
Johansen). Cette écriture admet une représentation VAR(p) équivalente **en niveaux**
(cf. Lütkepohl 2005, *New Introduction to Multiple Time Series Analysis*, pp. 269-304),
directement exploitable pour prévoir `yt+h` par récursion à partir des `p` dernières
observations réelles — c'est cette représentation (`var_rep` dans statsmodels) qui est
utilisée ci-dessous pour produire une prévision glissante à chaque date du test, avec des
paramètres figés à l'entraînement (aucune fuite du futur : seules des valeurs réelles
passées alimentent la récursion à chaque date).


```python
# ============================================================
# MOTEUR : par fold walk-forward, Johansen (train) -> VECM si r>0 -> prévision
# glissante à chaque date du test -> classe -> métriques, comparées à la référence ML.
# ============================================================
rows = []
t0 = time.time()

for k in range(CONFIG['n_wf_folds']):
    cut, nxt = FOLD_CUTS[k], FOLD_CUTS[k + 1]
    cut_date, nxt_date = all_dates[cut], all_dates[nxt - 1]
    train = system_df.loc[system_df.index < cut_date, VAR_COLS].dropna()
    if len(train) < CONFIG['min_train_rows']:
        print(f"Fold {k+1}: train trop court ({len(train)}) — ignoré.")
        continue

    sel = VAR(train).select_order(CONFIG['maxlags'])
    p_aic = sel.aic if sel.aic and sel.aic > 0 else 1

    johansen = coint_johansen(train.values, CONFIG['johansen_det_order'], CONFIG['maxlags'])
    trace_stat, trace_crit = johansen.lr1, johansen.cvt[:, 1]
    rank = 0
    for i in range(len(VAR_COLS)):
        if trace_stat[i] > trace_crit[i]:
            rank = i + 1
    rank = min(rank, len(VAR_COLS) - 1)

    if rank == 0:
        print(f"Fold {k+1}: Johansen ne détecte aucune cointégration (r=0) sur le train "
              f"-> VECM non estimable, fold ignoré (cf. VIX_VAR_MACRO pour ce cas).")
        continue

    k_ar_diff = max(p_aic - 1, 0)
    try:
        vecm_res = VECM(train, k_ar_diff=k_ar_diff, coint_rank=rank, deterministic='co').fit()
    except Exception as e:
        print(f"Fold {k+1}: [WARN] échec estimation VECM ({str(e)[:150]}) — ignoré.")
        continue

    coefs = vecm_res.var_rep
    p_eff = coefs.shape[0]
    const = vecm_res.const.ravel() if vecm_res.const is not None and vecm_res.const.size else np.zeros(len(VAR_COLS))
    print(f"\nFold {k+1}: train={len(train)}j | r={rank} | p_AIC={p_aic} (k_ar_diff={k_ar_diff}) "
          f"| p_eff(var_rep)={p_eff}")

    for h in CONFIG['horizons']:
        target, reg_r, thr = build_target(system_df['VIX'], h, cut)
        te_idx = target.index[(target.index >= cut_date) & (target.index <= nxt_date)]
        if len(te_idx) < CONFIG['min_test_rows']:
            continue

        y_true, y_pred = [], []
        for t in te_idx:
            pos = system_df.index.get_loc(t)
            if pos < p_eff - 1:
                continue
            y_hist = system_df[VAR_COLS].iloc[pos - p_eff + 1: pos + 1].values
            fc = var_level_forecast(coefs, const, y_hist, h)
            vix_fc = fc[VIX_IDX]; vix_now = system_df['VIX'].iloc[pos]
            ret_hat = vix_fc / vix_now - 1
            pred = classify_return(ret_hat, reg_r[t], thr)
            y_true.append(int(target.loc[t])); y_pred.append(pred)

        if len(y_true) < CONFIG['min_test_rows']:
            continue
        met = metrics(np.array(y_true), np.array(y_pred))
        rows.append({'fold': k + 1, 'horizon': h, 'coint_rank': rank, 'lag_p': p_eff,
                     'n_test': len(y_true), 'test_start': str(cut_date.date()),
                     'test_end': str(nxt_date.date()), **met})
        print(f"  h={h:2d}j n={len(y_true):4d} F1_dir={met['F1_dir']:.3f} "
              f"F1_UP_FORT={met['F1_UP_FORT']} F1_DOWN_FORT={met['F1_DOWN_FORT']}")

df_vecm = pd.DataFrame(rows)
df_vecm.to_csv(RESULTS_CSV, index=False)
print(f"\n[TERMINÉ] {len(df_vecm)} lignes (fold × horizon avec r>0) en {(time.time()-t0)/60:.1f}min "
      f"-> {RESULTS_CSV}")
```

```python
# ============================================================
# SYNTHÈSE : VECM vs référence ML établie
# ============================================================
if len(df_vecm):
    by_h = df_vecm.groupby('horizon')[['F1_dir', 'F1_UP_FORT', 'F1_DOWN_FORT', 'coint_rank']].mean().round(4)
    print("### Moyenne par horizon (folds où r>0) ###")
    print(by_h.to_string())

    overall_f1 = df_vecm['F1_dir'].mean()
    covered_folds = df_vecm['fold'].nunique()
    print(f"\nF1_dir moyen (tous folds/horizons avec r>0 confondus): {overall_f1:.4f} "
          f"(référence ML: {BASELINE_F1_DIR})")
    print(f"Folds avec cointégration détectée (r>0) : {covered_folds}/{CONFIG['n_wf_folds']}")

    if covered_folds == 0:
        print("\n[VERDICT] Aucun fold n'a de cointégration détectée sur son train — le VECM "
              "n'est testable sur aucune période. Voir VIX_VAR_BENCHMARK pour un test sans "
              "cette contrainte.")
    elif overall_f1 >= BASELINE_F1_DIR:
        print("\n[VERDICT] Le VECM égale ou dépasse la référence ML en F1_dir sur les "
              "périodes où une cointégration est détectée — résultat notable à creuser "
              "(mais échantillon probablement restreint, cf. folds couverts ci-dessus).")
    else:
        print(f"\n[VERDICT] Le VECM reste en-dessous de la référence ML (delta="
              f"{overall_f1 - BASELINE_F1_DIR:+.4f}) — la relation de cointégration "
              "captée par Johansen ne suffit pas, seule, à égaler un ensemble ML "
              "multi-features. Cohérent avec le reste du projet : aucune approche "
              "alternative testée n'a battu la référence GLOBAL RandomForest à ce jour.")

    try:
        with pd.ExcelWriter('VIX_VECM_report.xlsx', engine='xlsxwriter') as w:
            df_vecm.to_excel(w, 'Detail', index=False)
            by_h.reset_index().to_excel(w, 'Par_horizon', index=False)
            pd.DataFrame([{'F1_dir_moyen': overall_f1, 'F1_dir_reference_ML': BASELINE_F1_DIR,
                            'delta': overall_f1 - BASELINE_F1_DIR,
                            'folds_avec_cointegration': covered_folds,
                            'folds_total': CONFIG['n_wf_folds']}]).to_excel(w, 'Meta', index=False)
        print("\n[SAVE] VIX_VECM_report.xlsx")
    except Exception as e:
        print(f"[WARN Export] {e}")
else:
    print("Aucun résultat — aucun fold n'a de cointégration détectée (r=0 partout). "
          "Voir VIX_VAR_BENCHMARK pour un test sans cette contrainte.")
```
